```python
from debugbox import DOSBoxDebugger
from pathlib import Path
import time
from collections import Counter
import subprocess
```

```python
appname='/opt/homebrew/bin/dosbox-x'

def activate(app):
    subprocess.run(["osascript", "-e", f'tell application "System Events" to set frontmost of process "{app}" to true'])
def send_key_code(code, app=appname, modifier=None):
    subprocess.run(["osascript", "-e", f'tell application "{appname}" to activate'])
    mod = f" using {{{modifier}}}" if modifier else ""
    script = f'''
      tell application "System Events"
      tell process "{app}"
          key code {code}{mod}
      end tell
      end tell'''
    subprocess.run(["osascript", "-e", script])
```

```python
import re
ANSI_RE = re.compile(r'\x1B(?:[@-Z\\-_]|\[[0-?]*[ -/]*[@-~])')

def parse_mcb(raw):
    
    clean = ANSI_RE.sub('', raw)
    mcb = {}
    
    in_table = False
    
    for line in clean.splitlines():
        line = line.strip()
    
        if line.startswith("LOG: MCB Seg"):
            in_table = True
            continue
    
        if not in_table:
            continue
    
        # Stop when we hit the next section
        if line == "LOG: Upper memory:":
            break
    
        # Remove the LOG: prefix
        if line.startswith("LOG:"):
            line = line[4:].strip()
    
        # Ignore blank lines and END OF CHAIN
        if not line or "END OF CHAIN" in line:
            continue
    
        parts = line.split()
    
        # Need at least:
        # segment, size, PSP
        if len(parts) < 3:
            continue
    
        segment = parts[0]
        size = int(parts[1])
        psp = parts[2]
    
        # Remaining text is notes / filename
        rest = parts[3:]
    
        notes = None
        filename = None
    
        if rest:
            # "(DOS)" / "(free)" etc.
            if rest[0].startswith("("):
                notes = rest[0]
                rest = rest[1:]
    
            if rest:
                filename = rest[0]
    
        mcb[segment] = {
            "size": size,
            "psp": psp,
            "notes": notes,
            "filename": filename,
        }
    return mcb

def find_owner(hex_addr, mcb):
    addr = int(hex_addr, 16)

    for segment, block in mcb.items():
        start = int(segment, 16)
        end = start + (block["size"] // 16)

        if start <= addr < end:
            return {
                "segment": segment,
                "filename": block["filename"],
                "size": block["size"],
                "start": start,
                "end": end,
            }

    return None
```

```python
def get_logcpu_file():
    p = Path('/Users/robmaxwell/projects/f19/LOGCPU.TXT')
    text = p.read_text(errors="replace")
    return text

def get_pairs(text):
    pairs = [
        line.strip()
        for line in text.splitlines()
        if ":" in line
    ]
    counts = Counter(pairs)
    return counts

def get_transitions(text):
    pairs = [
        line.strip()
        for line in text.splitlines()
        if ":" in line
    ]
    transitions = Counter(
        zip(pairs[:-1], pairs[1:])
    )
    return transitions
```
